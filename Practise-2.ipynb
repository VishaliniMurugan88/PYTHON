```python
#Factorial
a=int(input("Enter number"))
p=1
for i in range(a,1,-1):
   p*=i
print("Factorial :",p)
```

```python
#Fibonacci series => sequence of numbers which each number is the sum of previous number.
a=-1
b=1
n=int(input("Enter number of terms:"))
print("Fibonacci series:",end=" ") 
for i in range(n):
    c=a+b
    a=b
    b=c
    print(c,end=" ")
```

```python
#Count digits
n=int(input("Enter numbers:"))
s=0
while n:
    s+=1
    n//=10
print("Count digits:",s)    
```

```python
#Sum of digits
n=int(input("Enter number:"))
s=0
while n:
    s+=n%10
    n//=10
print("sum of numbers:",s)   
```

```python
#Reverse number
n=int(input("Enter number:"))
s=0
while n:
    s=s*10+n%10
    n//=10
print("Reversed number:",s)    
```

```python
#Leap year
y=int(input("Enter the year:"))
if((y%4==0)and(y%100!=0)):
    print("Leap year")
else:
    print("Not a leap year")
```

```python
num=int(input("Enter the ID:"))
for i in range(2,num):
    if(num%i==0):
        print("Not a prime number")

    else:
        print("Prime number")
```
