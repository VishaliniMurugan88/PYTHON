```python
num=int(input("Enter the ID:"))
for i in range(2,num):
    if(num%i==0):
        print("Not a prime number")

    else:
        print("Prime number")
```

```python
def factorial(n):
    if(n==0)or(n==1):
        return 1
    else:
        return n*factorial(n-1)
n=int(input("Enter a num:")) 
print("factorial:",factorial(n))
```
