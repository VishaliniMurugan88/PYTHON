```python
a=0
b=1
for i in range(10):
    print(a,end="")
    c=a+b
    a=b
    b=c
```

```python
3.Write a Python function to check if a given string is a palindrome.
Expected Output (for input "madam"):
madam is a palindrome

```

```python
text=str(input("Enter the text:"))
rev=""
for i in text:
    rev=i+rev
if(text==rev):
    print("Palindrome")
else: 
    print("Not a Palindrome")
```
